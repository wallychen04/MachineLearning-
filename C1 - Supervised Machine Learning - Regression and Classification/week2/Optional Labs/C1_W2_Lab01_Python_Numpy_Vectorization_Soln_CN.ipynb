# 3 向量
<a name="toc_40015_3.1"></a>
## 3.1 概念
<img align="right" src="./images/C1_W2_Lab04_Vectors.PNG" style="width:340px;" >在本课程中，向量是*有序的数字数组*。记号上，向量用小写加粗字母表示，如 $\mathbf{x}$。向量的所有元素类型相同——一个向量不会既含字符又含数字。数组中元素的个数通常称为*维度*（dimension），不过数学家可能更偏好*秩*（rank）一词。图中所示向量的维度为 $n$。向量的元素可以通过索引引用。数学语境中索引通常从 1 到 n；而在计算机科学和这些实验中，索引通常从 0 到 n-1。记号上，单独引用向量元素时，索引用下标表示，例如向量 $\mathbf{x}$ 的第 $0$ 个元素写作 $x_0$。注意这里的 x 不加粗。


<a name="toc_40015_3.2"></a>
## 3.2 NumPy 数组

NumPy 的基本数据结构是可索引的、n 维的*数组*（array），其中所有元素类型相同（`dtype`）。你马上会注意到，“维度”一词在这里被复用了：上面它指向量中元素的个数，而这里维度指的是数组索引的个数。一维（1-D）数组只有一个索引。在课程一中，我们将用 NumPy 一维数组来表示向量。

 - 一维数组，形状 (n,)：n 个元素，索引从 [0] 到 [n-1]
 

<a name="toc_40015_3.3"></a>
## 3.3 创建向量


NumPy 的数据创建例程通常以对象的形状（shape）作为第一个参数。它可以是一个单值（生成一维结果），也可以是一个指定结果形状的元组 (n,m,...)。下面是用这些例程创建向量的例子。

```python
# NumPy 例程：分配内存并用值填充数组
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

有些数据创建例程不接受形状元组：

```python
# NumPy 例程：分配内存并用值填充数组，但不接受形状作为输入参数
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

也可以手动指定具体数值。

```python
# NumPy 例程：分配内存并用用户指定的值填充
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

以上都创建了一个含 4 个元素的一维向量 `a`。`a.shape` 返回各维度大小。这里 a.shape 为 `(4,)`，表示一个含 4 个元素的一维数组。

## 3.4 向量运算
让我们探索一些向量运算。
<a name="toc_40015_3.4.1"></a>
### 3.4.1 索引
向量元素可以通过索引（indexing）和切片（slicing）访问。NumPy 提供了一整套非常完善的索引和切片功能，这里我们只探索课程所需的基础内容，更多细节参考 [Slicing and Indexing](https://NumPy.org/doc/stable/reference/arrays.indexing.html)。
**索引**是指按元素在数组中的位置引用*单个元素*。
**切片**是指按索引从数组中取出一*部分*元素。
NumPy 的索引从 0 开始，所以向量 $\mathbf{a}$ 的第 3 个元素是 `a[2]`。

```python
# 一维向量的索引操作
a = np.arange(10)
print(a)

# 访问一个元素
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# 访问最后一个元素；负数索引从末尾往前数
print(f"a[-1] = {a[-1]}")

# 索引必须在向量范围内，否则会报错
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

<a name="toc_40015_3.4.2"></a>
### 3.4.2 切片
切片用三个值（`start:stop:step`，起始:终止:步长）生成一组索引，只写其中一部分值也是合法的。用例子来解释最直观：

```python
# 向量切片操作
a = np.arange(10)
print(f"a         = {a}")

# 访问连续 5 个元素（start:stop:step）
c = a[2:7:1];     print("a[2:7:1] = ", c)

# 每隔两个取一个，共访问 3 个元素
c = a[2:7:2];     print("a[2:7:2] = ", c)

# 访问索引 3 及之后的所有元素
c = a[3:];        print("a[3:]    = ", c)

# 访问索引 3 之前的所有元素
c = a[:3];        print("a[:3]    = ", c)

# 访问所有元素
c = a[:];         print("a[:]     = ", c)
```

<a name="toc_40015_3.4.3"></a>
### 3.4.3 单向量运算
有许多只涉及单个向量的实用运算。

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# 对 a 的各元素取负
b = -a 
print(f"b = -a        : {b}")

# 对 a 的所有元素求和，返回一个标量
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
```

### 3.4.4 向量-向量逐元素运算
NumPy 的大多数算术、逻辑和比较运算都适用于向量，它们按逐元素的方式工作。例如
$$ \mathbf{a} + \mathbf{b} = \sum_{i=0}^{n-1} a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
```

当然，要正确进行这类运算，两个向量的大小必须相同：

```python
# 试一下大小不匹配的向量运算
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

### 3.4.6 向量-向量点积
点积（dot product）是线性代数和 NumPy 的基石运算，本课程中会大量使用，务必理解透彻。点积如下图所示。

<img src="./images/C1_W2_Lab04_dot_notrans.gif" width=800> 

点积把两个向量中的对应值逐元素相乘，然后把结果求和。
向量点积要求两个向量的维度相同。

下面我们自己实现一个点积版本：

**使用 for 循环**实现一个函数，返回两个向量的点积。给定输入 $a$ 和 $b$，函数返回：
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
假设 `a` 和 `b` 形状相同。

```python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# 一维测试
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

注意，点积应返回一个标量值。

让我们用 `np.dot` 做同样的运算。

```python
# 一维测试
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

从上面可以看到，一维情形的结果与我们自己的实现一致。

### 3.4.7 速度的需求：向量化 vs for 循环
我们使用 NumPy 库是因为它能提升速度和内存效率。来演示一下：

```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

可以看到，在这个例子中向量化带来了巨大的速度提升。这是因为 NumPy 能更好地利用底层硬件的数据并行能力：GPU 和现代 CPU 都实现了单指令多数据（SIMD）流水线，可以并行发起多个运算。在数据集通常很大的机器学习中，这一点至关重要。

## 4.3 创建矩阵
创建一维向量的那些函数同样可以创建二维或 n 维数组。下面是一些例子。


下面提供了形状元组以生成二维结果。注意 NumPy 如何用方括号区分每个维度；还可以注意到，NumPy 打印时每行占一行。

## 4.4 矩阵运算
让我们探索一些矩阵运算。

<a name="toc_40015_4.4.1"></a>
### 4.4.1 索引


矩阵有第二个索引，两个索引描述 [行, 列]。访问可以返回单个元素，也可以返回一行/一列。见下：

```python
# 矩阵的索引操作
a = np.arange(6).reshape(-1, 2)   #reshape is a convenient way to create matrices
print(f"a.shape: {a.shape}, \na= {a}")

# 访问一个元素
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

# 访问一行
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

最后一个例子值得特别注意：只指定行来访问矩阵时，返回的是一个*一维向量*。

**Reshape（变形）**
上一个例子使用 [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) 来改变数组形状。
`a = np.arange(6).reshape(-1, 2) `
这行代码先创建了一个含 6 个元素的*一维向量*，然后用 reshape 命令把它变形成*二维*数组。它也可以写成：
`a = np.arange(6).reshape(3, 2) `
得到同样的 3 行 2 列数组。
-1 这个参数告诉例程：根据数组总大小和列数自动计算行数。


<a name="toc_40015_4.4.2"></a>
### 4.4.2 切片
切片用三个值（`start:stop:step`）生成一组索引，只写其中一部分值也是合法的。用例子来解释最直观：

```python
# 二维切片操作
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

# 访问连续 5 个元素（start:stop:step）
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

# 在两行中各访问连续 5 个元素（start:stop:step）
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# 访问所有元素
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# 访问某一行的所有元素（非常常用的写法）
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# 等同于
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
