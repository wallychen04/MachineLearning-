# 不计分实验：用 Scikit-Learn 实现逻辑回归




## 目标
在本实验中，你将：
- 使用 scikit-learn 训练一个逻辑回归模型


## 数据集
让我们使用和之前相同的数据集。

```python
import numpy as np

X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
```

## 拟合模型

下面的代码从 scikit-learn 导入[逻辑回归模型](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html#sklearn.linear_model.LogisticRegression)。调用 `fit` 函数即可在训练数据上拟合该模型。

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X, y)
```

## 进行预测

调用 `predict` 函数，就可以看到该模型做出的预测。

```python
y_pred = lr_model.predict(X)

print("Prediction on training set:", y_pred)
```

## 计算准确率

调用 `score` 函数，就可以计算该模型的准确率。

```python
print("Accuracy on training set:", lr_model.score(X, y))
```
