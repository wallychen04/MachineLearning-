### 为什么需要非线性激活？
<img align="left" src="./images/C2_W2_ReLU_Graph.png"     style=" width:250px; padding: 10px 20px; " > 图中所示函数由若干线性片段组成（分段线性）。斜率在每个线性段内保持不变，并在转折点处突变。在转折点处，一个新的线性函数被加进来，与已有函数相加后产生新的斜率。新函数在转折点处加入，但在该点之前不对输出产生任何贡献。非线性激活函数正是负责在转折点之前（有时也包括之后）把输入“关掉”的。下面的练习会给出一个更具体的例子。

 
这个练习的目的是体会 ReLU 的非线性行为如何提供了“在需要之前把函数关掉”的能力。让我们看看这个例子中它是如何工作的。
<img align="right" src="./images/C2_W2_ReLU_Plot.png"     style=" width:600px; padding: 10px 20px;"> 
右侧的图展示了第一层各单元的输出。
从最上方开始，单元 0 负责标记为 1 的第一段。图中同时画出了线性函数 $z$ 和经过 ReLU 之后的函数 $a$。可以看到，ReLU 在区间 [0,1] 之后把函数截断了。这很重要——它防止了单元 0 干扰后面的段。

单元 1 负责第二段。这里 ReLU 让该单元在 x 到达 1 之前保持“安静”。由于第一个单元没有贡献，单元 1 的斜率 $w^{[1]}_1$ 就等于目标线的斜率。偏置必须调整到使输出在 x 到达 1 之前保持为负。注意单元 1 的贡献还延伸到了第三段。

单元 2 负责第三段。ReLU 同样在 x 到达合适值之前把输出归零。该单元的斜率 $w^{[1]}_2$ 必须设置成使单元 1 与单元 2 之和具有期望的斜率。偏置同样需要调整，使输出在 x 到达 2 之前保持为负。

正是 ReLU 激活的“关闭/禁用”特性，使模型能够把线性片段拼接起来，对复杂的非线性函数建模。
