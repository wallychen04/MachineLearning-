## 导数的非正式定义

导数的正式定义涉及极限和“趋近于零”之类的说法，听起来有点吓人，但它的核心思想其实简单得多。

函数的导数描述的是：当输入变量发生微小变化时，函数的输出会如何变化。

让我们以代价函数 $J(w)$ 为例：代价 $J$ 是输出，$w$ 是输入变量。
我们给“微小变化”起个名字 *epsilon*（$\epsilon$）。用希腊字母是因为数学上传统上用 *epsilon*（$\epsilon$）或 *delta*（$\Delta$）表示很小的值。你可以把它理解为 0.001 之类的某个小值。

$$
\begin{equation}
\text{if } w \uparrow \epsilon \text{ causes }J(w) \uparrow \text{by }k \times \epsilon \text{ then}  \\
\frac{\partial J(w)}{\partial w} = k \tag{1}
\end{equation}
$$

这说的是：如果你让函数 $J(w)$ 的输入变化一点点，而输出变化了那一小点的 $k$ 倍，那么 $J(w)$ 的导数就等于 $k$。

来试一下。我们来看函数 $J(w) = w^2$ 在 $w=3$、$\epsilon = 0.001$ 处的导数

````python
J = (3)**2
J_epsilon = (3 + 0.001)**2
k = (J_epsilon - J)/0.001    # difference divided by epsilon
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k:0.6f} ")
````

我们把输入值增大了一点点（0.001），导致输出从 9 变为 9.006001，增量是输入增量的 6 倍。对照上面的 (1)，这说明 $k=6$，所以 $\frac{\partial J(w)}{\partial w} \approx 6$。如果你学过微积分，就知道用符号写法是 $\frac{\partial J(w)}{\partial w} = 2 w$，代入 $w=3$ 正好是 6。我们上面的计算不恰好是 6，因为要完全精确，$\epsilon$ 必须[无穷小](https://www.dictionary.com/browse/infinitesimally)——真正、真正地小。这就是为什么我们用 $\approx$ 或 ~= 而不是 =。让我们看看把 $\epsilon$ 取得更小会怎样。

````python
J = (3)**2
J_epsilon = (3 + 0.000000001)**2
k = (J_epsilon - J)/0.000000001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

随着 $\epsilon$ 缩小，结果越来越接近精确的 6。你可以自己试着把值调得更小。

## 求符号导数
在反向传播中，我们常常需要知道简单函数在任意输入值处的导数。换句话说，我们想要的是“符号”导数，而不是“算术”导数。符号导数的例子如上面 $J(w) = w^2$ 的导数 $\frac{\partial J(w)}{\partial w} = 2 w$。有了符号导数，你就能求出任意输入值 $w$ 处的导数值。

如果你上过微积分课，就熟悉数学家们为求导总结的大量[求导法则](https://en.wikipedia.org/wiki/Differentiation_rules#Power_laws,_polynomials,_quotients,_and_reciprocals)。而事实证明，这个过程已经被符号求导程序自动化了。Python 中的例子是 [SymPy](https://www.sympy.org/en/index.html) 库。让我们看看怎么用它。

### $J = w^2$
定义 Python 变量及其符号名。

````python
J, w = symbols('J, w')
````

定义并打印表达式。注意 SymPy 会生成一个 [LaTeX](https://en.wikibooks.org/wiki/LaTeX/Mathematics) 字符串，渲染出美观易读的公式。

````python
J=w**2
J
````

用 SymPy 的 `diff` 求 $J$ 对 $w$ 的导数。注意结果与我们之前的例子一致。

````python
dJ_dw = diff(J,w)
dJ_dw
````

通过把数值“代入”符号值，在几个点上求导数的值。在第一个例子中，$w$ 被替换为 $2$。

````python
dJ_dw.subs([(w,2)])    # derivative at the point w = 2
````

````python
dJ_dw.subs([(w,3)])    # derivative at the point w = 3
````

````python
dJ_dw.subs([(w,-3)])    # derivative at the point w = -3
````

## $J = 2w$

````python
w, J = symbols('w, J')
````

````python
J = 2 * w
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,-3)])    # derivative at the point w = -3
````

与算术计算的结果对比一下

````python
J = 2*3
J_epsilon = 2*(3 + 0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

对函数 $J=2w$，很容易看出：无论 $w$ 的初值是多少，$w$ 的任何变化都会使输出 $J$ 产生 2 倍的变化。NumPy 和算术结果都印证了这一点。

## $J = w^3$

````python
J, w = symbols('J, w')
````

````python
J=w**3
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,2)])   # derivative at the point w=2
````

与算术计算的结果对比一下

````python
J = (2)**3
J_epsilon = (2+0.001)**3
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

## $J = \frac{1}{w}$

````python
J, w = symbols('J, w')
````

````python
J= 1/w
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,2)])
````

与算术计算的结果对比一下

````python
J = 1/2
J_epsilon = 1/(2+0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````
